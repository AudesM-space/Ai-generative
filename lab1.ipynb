```python
x = torch.tensor(3.0)
W = torch.tensor(4.0, requires_grad=True)
b = torch.tensor(6.0, requires_grad=True)

y = W * x + b
loss = (y - 10.0) ** 2
```

```python
loss
```

```python
loss.backward()
```

```python
print(W.grad.item())
```

```python
print(b.grad.item())
```

```python
dW = torch.autograd.grad(loss,W)
```
